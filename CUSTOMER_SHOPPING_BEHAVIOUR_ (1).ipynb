## 3. ***Data Wrangling***

### Data Wrangling Code

```python
# Write your code to make your dataset analysis ready.
# Step 1: Create a Copy of the Dataset
# Create a copy of the original dataset
df_copy = df.copy()

# Step 2: Check Duplicate Records
# Check for duplicate rows
duplicate_count = df_copy.duplicated().sum()

print("Duplicate Records:", duplicate_count)
#Step 3: Remove Duplicate Records (If Any)
# Remove duplicate rows
df_copy.drop_duplicates(inplace=True)

print("Dataset Shape after Removing Duplicates:", df_copy.shape)
#Step 4: Check Missing Values
# Check missing values in each column
df_copy.isnull().sum()

#Step 5: Handle Missing Values

# Fill missing values in Review Rating with median

df_copy["Review Rating"] = df_copy["Review Rating"].fillna(df_copy["Review Rating"].median())

#Step 6: Verify Missing Values
# Verify that missing values have been handled

df_copy.isnull().sum()

#Step 7: Final Dataset Shape
# Display final dataset shape
print("Final Dataset Shape:", df_copy.shape)
```

### What all manipulations have you done and insights you found?

During the data wrangling process, the following data cleaning steps were performed:

Created a copy of the original dataset (df_copy) to preserve the raw data.
Checked for duplicate records using duplicated().sum().
Found 0 duplicate records, indicating that all customer records are unique.
Checked for missing values in all columns.
Identified 37 missing values in the Review Rating column.
Replaced the missing values in the Review Rating column with the median value to maintain the data distribution and avoid the effect of outliers.
Verified that all missing values were successfully handled.
Confirmed the final dataset shape as 3,900 rows × 18 columns, showing that no records were removed during the cleaning process.

**Insights Found**
The dataset is clean and well-structured, with no duplicate records.
Only one column (Review Rating) contained missing values, accounting for approximately 0.95% of the dataset.
Using the median to fill missing values preserved the integrity of the numerical data.
Since no rows were deleted, the dataset retained all 3,900 customer records, ensuring maximum information for further analysis.
The cleaned dataset is now ready for Exploratory Data Analysis (EDA) and can be used to identify customer shopping patterns, purchasing behavior, and business insights.

## **5. Solution to Business Objective**

#### What do you suggest the client to achieve Business Objective ?
Explain Briefly.

Solution to Business Objective
To help the client grow revenue, improve customer retention, and operate more efficiently, I suggest the following actions based on the EDA findings:

Focus inventory and promotion on top-performing categories. Categories with higher purchase frequency and higher average purchase amount (from Charts 3 and 6) should get priority shelf space, ad spend, and stock levels, since they already have proven demand.
Invest in growing the subscription base. Subscription adoption is currently low (Chart 9). The client should introduce stronger incentives — free shipping, early access to sales, or subscriber-only discounts — to convert one-time buyers into recurring subscribers, which stabilizes revenue over time.

Use discounts and promo codes strategically, not as a default. A large share of purchases already involve a discount or promo code (Charts 12 and 13). Rather than broad, always-on discounting (which erodes margin), the client should shift toward targeted promotions — e.g., win-back offers for lapsed customers or first-purchase incentives — while encouraging full-price purchases from loyal customers through non-discount perks.
Streamline payment and shipping options. Since a small number of payment methods and shipping types account for most usage (Charts 8 and 10), the client can simplify checkout by promoting these preferred options, potentially reducing operational costs from maintaining rarely-used alternatives.

Plan around seasonal demand. Purchases show mild seasonal variation (Chart 7). The client should align inventory build-up and marketing campaigns with the higher-demand season(s) and design targeted promotions to lift sales in the slower season.

Monitor and act on review ratings. While most ratings are moderate-to-high (Chart 11), the client should set up a process to flag and follow up on low-rated purchases quickly (e.g., customer service outreach), since unaddressed dissatisfaction is a direct churn risk.
Use purchase amount and previous-purchase patterns for segmentation. Since numerical variables show weak correlation with each other (Chart 14) but purchase amount, review rating, and previous purchases together reveal customer behavior patterns (Chart 15), the client can build simple customer segments (e.g., high-value/low-frequency vs. low-value/high-frequency) to personalize marketing rather than treating all customers the same.

Together, these steps target the two levers that move the business objective forward: increasing revenue per customer (via category focus, smarter discounting, and segmentation) and increasing customer retention (via subscriptions, seasonal planning, and proactive service on low ratings).
